```python
# Final Fertilizer Recommendation Profiles

fertilizer_profiles = df.groupby("Fertilizer Name")[
    [
        "Nitrogen",
        "Potassium",
        "Phosphorous",
        "Temparature",
        "Humidity",
        "Moisture"
    ]
].agg(["mean", "min", "max"]).round(2)

display(fertilizer_profiles)
```

```python
# Fertilizer Recommendation Profiles

fertilizer_profiles = {
    "Urea": {
        "N": 46,
        "P": 0,
        "K": 0,
        "use": "Nitrogen requirement"
    },
    "DAP": {
        "N": 18,
        "P": 46,
        "K": 0,
        "use": "Nitrogen + Phosphorus requirement"
    },
    "10-26-26": {
        "N": 10,
        "P": 26,
        "K": 26,
        "use": "Balanced NPK with higher Potassium"
    },
    "14-35-14": {
        "N": 14,
        "P": 35,
        "K": 14,
        "use": "Higher Phosphorus requirement"
    },
    "17-17-17": {
        "N": 17,
        "P": 17,
        "K": 17,
        "use": "Balanced NPK requirement"
    },
    "20-20": {
        "N": 20,
        "P": 20,
        "K": 0,
        "use": "Nitrogen + Phosphorus requirement"
    },
    "28-28": {
        "N": 28,
        "P": 28,
        "K": 0,
        "use": "Higher Nitrogen + Phosphorus requirement"
    }
}

print("Fertilizer profiles created successfully.")
print("Total fertilizers:", len(fertilizer_profiles))

for name, profile in fertilizer_profiles.items():
    print(
        f"{name}: "
        f"N={profile['N']}%, "
        f"P={profile['P']}%, "
        f"K={profile['K']}%"
    )
```

```python
# FINAL FERTILIZER RECOMMENDATION ENGINE

def recommend_fertilizers(nitrogen, phosphorus, potassium, soil_type, crop_type, top_n=3):

    recommendations = []

    # Nutrient requirement levels
    n_need = max(0, 40 - nitrogen)
    p_need = max(0, 40 - phosphorus)
    k_need = max(0, 20 - potassium)

    # Soil-based adjustment
    soil_bonus = {
        "Sandy": {"N": 1.20, "P": 1.00, "K": 1.15},
        "Loamy": {"N": 1.00, "P": 1.00, "K": 1.00},
        "Clayey": {"N": 0.90, "P": 1.00, "K": 0.95},
        "Black": {"N": 0.95, "P": 1.00, "K": 1.00},
        "Red": {"N": 1.10, "P": 1.05, "K": 1.00}
    }

    soil_factor = soil_bonus.get(
        soil_type,
        {"N": 1.0, "P": 1.0, "K": 1.0}
    )

    for fertilizer, profile in fertilizer_profiles.items():

        n_score = n_need * profile["N"] * soil_factor["N"]
        p_score = p_need * profile["P"] * soil_factor["P"]
        k_score = k_need * profile["K"] * soil_factor["K"]

        total_score = n_score + p_score + k_score

        # Crop compatibility bonus
        crop_bonus = 0

        if crop_type in ["Paddy", "Wheat", "Maize", "Barley"]:
            if fertilizer in ["Urea", "DAP", "28-28"]:
                crop_bonus = 15

        elif crop_type in ["Pulses", "Millets", "Oil seeds"]:
            if fertilizer in ["17-17-17", "10-26-26", "14-35-14"]:
                crop_bonus = 15

        elif crop_type in ["Cotton", "Sugarcane", "Tobacco"]:
            if fertilizer in ["10-26-26", "17-17-17", "28-28"]:
                crop_bonus = 15

        final_score = total_score + crop_bonus

        recommendations.append({
            "Fertilizer": fertilizer,
            "Score": round(final_score, 2),
            "N Contribution": profile["N"],
            "P Contribution": profile["P"],
            "K Contribution": profile["K"],
            "Reason": profile["use"]
        })

    recommendations = sorted(
        recommendations,
        key=lambda x: x["Score"],
        reverse=True
    )

    return recommendations[:top_n]


# TEST CASE
result = recommend_fertilizers(
    nitrogen=20,
    phosphorus=20,
    potassium=10,
    soil_type="Sandy",
    crop_type="Wheat",
    top_n=3
)

display(pd.DataFrame(result))
```
